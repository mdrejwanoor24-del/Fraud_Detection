# Data Pre-processing
- Data type correction
- Duplicate Handling
- Remove irrelevent columns
- Save Cleaned Data

```python
#imports
import pandas as pd
import numpy as np
```

```python
#load data
DATA_PATH=r"C:\Users\DELL\OneDrive\Documents\Fraud_Detection_Project\Data\Fraud_Analysis_Dataset.csv"
data=pd.read_csv(DATA_PATH)
data.head()
```

```python
#data backup
df=data.copy()
data.info()
```

```python
#meta data
print("Number of rows:",df.shape[0])
print("Number of columns:",df.shape[1])
all_columns=df.columns.to_list()
num_columns=df.select_dtypes(include="number").columns.to_list()
cat_columns=df.select_dtypes(include="object").columns.to_list()
print("\n All Columns :\n",all_columns)
print("\n num columns :\n",num_columns)
print("\n cat columns:\n",cat_columns)
print("\n data types:\n",df.dtypes)
```

```python
#duplicate details
print("Duplicate Details:",df.duplicated().sum())
print("Duplicate rows:\n")
df[df.duplicated()]
```

```python
#drop irrelevent column
df.drop(columns=["nameOrig","nameDest"],axis=1,inplace=True)
```

```python
#save cleaned data
CLEANED_DATA_PATH=r"C:\Users\DELL\OneDrive\Documents\Fraud_Detection_Project\Data\cleaned_data.csv"
df.to_csv(CLEANED_DATA_PATH,index=False)
```
